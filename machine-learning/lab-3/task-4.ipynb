```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("../../datasets/2019.csv")
```

```python
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Country or region"] = df["Country or region"].astype("category")
df["Overall rank"] = df["Overall rank"].astype(int)
df.info()
```

```python
df.columns = df.columns.str.strip()
print("Cleaned Column Names:", list(df.columns))
```

```python
top10_happiest = df.sort_values(by="Score", ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x="Score", y="Country or region", data=top10_happiest, palette="viridis")
plt.title("Top 10 Happiest Countries (2019)", fontsize=14, fontweight="bold")
plt.xlabel("Happiness Score", fontsize=12)
plt.ylabel("Country or Region", fontsize=12)
plt.xlim(0, 10)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="GDP per capita", y="Score", data=df, color="royalblue", s=60, alpha=0.8)
plt.title("GDP per Capita vs Happiness Score", fontsize=14, fontweight="bold")
plt.xlabel("GDP per Capita", fontsize=12)
plt.ylabel("Happiness Score", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Healthy life expectancy", y="Score", data=df, color="forestgreen", s=60, alpha=0.8)
plt.title("Healthy Life Expectancy vs Happiness Score", fontsize=14, fontweight="bold")
plt.xlabel("Healthy Life Expectancy", fontsize=12)
plt.ylabel("Happiness Score", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
