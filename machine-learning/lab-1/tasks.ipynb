```python
# E3

name = input("Enter your name: ")
separator_index = name.index(".")

first_name = name[0:separator_index]
last_name = name[separator_index+1:]

print(f"First Name = {first_name}")
print(f"Last Name = {last_name}")

sentence = input("Enter sentence to split: ")
print(sentence.split(" "))
```

```python
# E4: list

numbers = [1, 2, 3, 4, 5]
print(f"before appending {numbers}")
numbers.append(6)
print(f"after appending {numbers}\n")

numbers.extend([7, 8, 9])
print(f"after appending {numbers}\n")

print(numbers[:3])
print(numbers[-3:])
print(numbers[2:])
```

```python
# E5: dict
bio = {
    "name": "Naveed",
    "age": 30,
    "address": "xyz"
}

print(bio)
bio["age"] = 23
bio["email"] = "xyz@gmail.com"

print(bio.get("email"))
print(bio["email"])

bio.pop("address")
print(bio)
```
