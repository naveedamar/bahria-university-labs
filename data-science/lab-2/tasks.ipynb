```python
# E1: Calculating Standard Deviation and Variance of annual_rent_usd of rentals.csv

rental_db = pd.read_csv("../../datasets/rentals.csv")
print(rental_db.dtypes)

rental_mean = rental_db["annual_rent_usd"].mean()
print()

# sample variance, sum(x-mean)^2 / n-1
rental_variance = sum((rental_db["annual_rent_usd"] - rental_mean) ** 2) / (len(rental_db["annual_rent_usd"]) - 1)
print(f"Variance: {round(rental_variance)}")

# standard deviation, sqrt(variance)
rental_stdev = math.sqrt(rental_variance)
print(f"Standard Deviation: {round(rental_stdev)}")
```

```python
# E3: Palindrome checker

def check_palindrome(string):
    string = string.lower()
    if string == string[::-1]:
        return True
    else:
        return False
print(check_palindrome("Machine Learning"))
print(check_palindrome("BOB"))
```

```python
# E4: fibonaci series checker

def fib_series_checker(target):
    a = 0
    b = 1
    while a <= target:
        if a == target:
            return True
        a, b = b, a + b
    return False


fib_to_check = int(input("Enter number to check if it is in fibonaci series: "))
if fib_series_checker(fib_to_check):
    print(f"{fib_to_check} is in the Fibonacci series.")
else:
    print(f"{fib_to_check} is not in the Fibonacci series.")
```
